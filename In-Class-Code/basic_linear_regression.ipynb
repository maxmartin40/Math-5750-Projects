Import necessary packages

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

pd.set_option("display.precision", 3)
```

Import data and view dataset for formatting and missing values.

```python
df = sns.load_dataset("mpg")
print(df.shape)
df.head()
```

```python
df.isna().sum()
```

Drop all rows with no data and list the remaining number of entries $N$.

```python
df = df.dropna(subset=["horsepower"])
print(f"N = {len(df)} cars remain")
```

We can print summary statistics to visualize the distribution of the dataset.

```python
df[["horsepower", "mpg", "weight"]].describe()
```

Make a scatter plot with the data.

```python
fig, ax = plt.subplots(figsize=(5.5, 4))
ax.scatter(df.horsepower, df.mpg, s=14, alpha=0.6)
ax.set_xlabel("horsepower"); ax.set_ylabel("mpg")
ax.set_title("Auto MPG: 392 cars")
plt.show()
```

Now we begin to run the linear regression on this dataset.

It is noted that $X$ must be in the shape (N,1) which is why there are double brackets in its definition.

```python
X = df[["horsepower"]]     # DataFrame, shape (392, 1)
y = df["mpg"]              # Series,    shape (392,)
print("X", X.shape, " y", y.shape)

model = LinearRegression().fit(X, y)
print(f"intercept = {model.intercept_:.4f}")
print(f"slope     = {model.coef_[0]:.4f}")
```

Now that we have our model, we want to determine how effective it is by calculating different error types.

```python
yhat = model.predict(X)

print(f"sklearn:  RMSE = {np.sqrt(mean_squared_error(y, yhat)):.4f}"
      f"   MAE = {mean_absolute_error(y, yhat):.4f}"
      f"   R^2 = {r2_score(y, yhat):.4f}")
```

Interpretation: RMSE (root mean squared error) and MAE (mean absolute error) tell us that the model is normally off by 4.89 MPG and 3.83 MPG respectively. The $R^2$ value tells us that about 61% of the variance of the dataset is explained by our linear regression model.

Now we want to plot our prediction model so that we can visually see how well it does against our data.

```python
grid = pd.DataFrame({"horsepower": np.linspace(df.horsepower.min(), df.horsepower.max(), 200)})

fig, ax = plt.subplots(figsize=(5.5, 4))
ax.scatter(df.horsepower, df.mpg, s=14, alpha=0.5, label="data")
ax.plot(grid.horsepower, model.predict(grid), color="crimson", lw=2,
        label="linear regression model")
ax.set_xlabel("horsepower"); ax.set_ylabel("mpg"); ax.legend()
plt.show()
```

We can see that the model doesn't do a *great* job explaining our data in this example. We will explore other methods to better explain our model in another class.
